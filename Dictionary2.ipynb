## case study

1. Create a dictionary named `student` with the following keys:
   name, age, course

2. Store multiple students' data using lists as values.

3. Add one new student details to the dictionary:

   • Append new name to name list  

   • Append new age to age list  

   • Append new course to course list  

4. Display all student details using dictionary keys.

5. Access and display:

   • First student's details using indexing

   • Newly added student's details using indexing





```python

#Store multiple students' data using lists as values.
student = {
    "name":   ["krishna", "ram", "gowrangi"],
    "age":    [21, 22, 23],
    "course": ["Data Science", "AI", "Machine Learning"]
}
print(student)

#Add one new student details to the dictionary:
student['name'].append("Gopika")
student['age'].append("28")
student["course"].append("cloudComputing")

print(student)


print("Name:", student["name"])
print("Age:", student["age"])
print("Course:", student["course"])

#First student's details using indexing


print("\nFirst Student Details")
print("Name:", student["name"][0])
print("Age:", student["age"][0])
print("Course:", student["course"][0])
```

# Case Study – Student Course Registration System using Set
Instructions:

Create a set named courses with course names: Python, SQL, Power BI, Python, Excel

Perform set operations: • Add course using add()
• Add multiple courses using update()
• Remove course using remove() and discard()

Display all courses and total number of courses using len()

```python

#Create a set named courses with course names

course = {"sql","Python","genAI"}
print(course)
print(type(course))

# Perform set operations:
# Add course using add() :
# Add multiple courses using update()
# Remove course using remove() and discard()

course.add("AI")
course.update(["DA","Advanced DAX"])


print(course)

course.discard("genAI")


print(course)

# Display total number of courses
print("Total Number of Courses:", len(["course"]))


# Dispaly all courses

print("All Courses :" ,course )
```
